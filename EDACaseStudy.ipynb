```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv('/content/purchase_data.csv')
```

```python
data
```

```python
#For one product price is too high. There was something wrong with this column name
#Product ID -> Purchase ID
```

```python
data = data.rename(columns={'Product_ID': 'Purchase_ID'})
```

```python
data
```

```python
#Information of the columns
# Unique values of Stay_In_Current_City_Years column
data.info()
```

```python
#Object column type -> The column which contains catagorical data -> Characters, strings,
```

```python
# Unique values of Stay_In_Current_City_Years column
data['Stay_In_Current_City_Years'].unique()

unique_values = data['Stay_In_Current_City_Years'].unique()
unique_values
```

```python
#Because of this 4+ category in my column, it is considering my column as object type.
```

```python
# In this column replace 4+ -> 4
data['Stay_In_Current_City_Years'].replace("4+","4",inplace=True)
```

```python
data
```

```python
data['Stay_In_Current_City_Years'].unique()
```

```python
#I know over here all values are of integer type -> Casting
data['Stay_In_Current_City_Years'] = data['Stay_In_Current_City_Years'].astype(float)
```

```python
data.info()
```

```python
#Checking for null values
data.isnull()
```

```python
#Finding the sum
data.isnull().sum().sum()
data.isnull().sum()
```

```python
#Now my task is to fill those empty places, values
#fillna()
#In these 2 columns Prod_Cat2 Prod_Cat3 most of my null values
```

```python
data[['Product_Category_2', 'Product_Category_3']] = data[['Product_Category_2', 'Product_Category_3']].fillna(0)
```

```python
data.isnull().sum()
```

```python
#Data visualization concepts
plt.boxplot(data['Purchase'])
```

```python
#Find outliers for my Purchase column.
sns.boxplot(y=data['Purchase'])
plt.show()
```

```python
Q1 = data['Purchase'].quantile(0.25)
Q3 = data['Purchase'].quantile(0.75)
```

```python
IQR = Q3-Q1
```

```python
print(IQR)
print(Q3)
print(Q1)
```

```python
#lower bound and upper bound
ll = Q1 - 1.5*IQR        #What ever values in my ll has to be removed.
ul = Q3 + 1.5 * IQR      #What ever values in my ul has to be removed.
```

```python
ll
ul
```

```python
#Give all the rows who are having outliers with respect to our purchase column.
```

```python
outliers = data[(data['Purchase']<(Q1-1.5*IQR)) | (data['Purchase']>(Q3+1.5*IQR))]
```
