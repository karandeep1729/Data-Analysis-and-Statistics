# How to find mean, median of values

```python
# Mean -> average
# data= [10,7,8,5,4]

#mean = sum of all values/ total number of values
#mean = 10+7+8+5+4/5 -> 6.8
#mean  = 6.8
```

```python
data= np.array([10,7,8,5,4])
```

```python
avg = np.mean(data)
avg
```

```python
# data = [10,5,4,23,65,78,12,1]
#Sort my data either in ascending or descending order
#sorted = [1,4,5,10,12,23,65,78]
#now look for middle value
#for even number of elements there will be 2 middle values
#[10,12]
#find their avg -> 10+12/2 => 22/2 -> 11 -> MEDIAN
```

```python
#Median ->
data = np.array([3,1,20,12,8,6,18,14])
med = np.median(data)
med
```
