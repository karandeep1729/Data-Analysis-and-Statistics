```python
new_data = {
    'Emp_Name':['Kabir','Mahesh','Phanendharnadh','Preethi','Maxime','Varsha'],
    'Emp_ID':[1021,1014,1045,1945,1123,1432],
    'Salary':[34000,45000,54000,39000,76000,59000],
    'Designation':['ML Engineer','Data Analyst','Research Analyst','AI Engineer','SDE','Business Analyst'],
    'Shift_timings':['9-5','10-7','11-8','9-5','12-9','8-4']
}
```

```python
emp_details = pd.DataFrame(new_data)
emp_details
```

```python
#I want to access designation
emp_details['Designation']   #Accessing particular column from our dataframe
```

```python
#loc function is used to access your column and rows by it's name
#iloc function is used to access your data by indexes.
```

```python
emp_details
```

```python
emp_details.loc[:, ['Emp_ID','Designation','Shift_timings']]
#When we are not aware of the indexes and we want to access our columns we use loc fxn
```

```python
emp_details.iloc[:,:4]  #Here we are accessing our data via indexing.
```

```python
emp_details.iloc[:,2:]
```
