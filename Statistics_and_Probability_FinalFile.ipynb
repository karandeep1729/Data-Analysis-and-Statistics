## **Outliers:** Extreme high values or extreme low values present in our data are known as outliers.

```python
# data =[50,60,90,100,45,75,85,2000]  #This 2000 is an Outlier (Extreme high value)
# ex = [5500,4800,7200,6000,4500,24]  #This 24 is also an Outlier.  (Extreme low value)
```

```python
# Outliers can affect the efficiency of the model by reducing it, that's why we clean our data
# by removing the outliers before building the model.

age = [18,25,30,45,33,67,-190,78,89,93,250,76,271]
print(age)
```

```python
import matplotlib.pyplot as plt
```

```python
# Boxplots are used to determine outliers present in our data
plt.boxplot(age)
plt.show()
```

image.png

```python
# We are aware that if in our data, any outliers are present
# they will be above Q3 or below Q1.
# Step1: Calculate Q1 and Q3.
import numpy as np
Q1 = np.percentile(age,25)
Q3 = np.percentile(age,75)
```

```python
print(Q1,Q3)
# I decide to remove all the values that are smaller than Q1 and greater than Q3.
#
```

```python
print(age)
# 18, 25, -190, 93, 250, 271
```

```python
# Step 2: Calculate IQR (Inter Quartile Range)
IQR = Q3 - Q1
print(IQR)
```

```python
# Step 3: Calculator lower_fence and upper_fence

lower_fence = Q1 - 1.5*IQR
upper_fence = Q3 + 1.5*IQR
```

```python
print(lower_fence, upper_fence)
```

```python
print(age)
```

```python
for x in age:
    if x < lower_fence or x > upper_fence:
        age.remove(x)
```

```python
plt.boxplot(age)
plt.show()
```

## **Skewness:** A concept, that is used to determine how the data is distributed.

```python
# Skewness ~=0 => Syymetric Data  (Normally Distributed Data)
# Skewness > 0 => Right-Skewed (Longer Right Tail) #Higher impact towards smaller values
# Skewness < 0 => Left-Skewed (Longer Left Tail)   #Higher impach towards larger values
```

## **Probability:**

It is a measure of how likely an event is going to occur.

1.  **Experiment:** Any process or task. Examples: Flipping a coin, rolling a dice, drawing a card from the deck.

2.  **Event:** Event is a specific or group of outcome from an experiment.

3.  **Simple Event:** Means 1 specific outcome. Ex: Flipping a coin and getting a tail.

4.  **Compound Event:** Means multiple outcomes. Rolling a dice and asking for even numbers (2,4,6).

5.  **Outcome:** It is like a result of single trial of experiment. Ex: When you are going to flip a coin, the outcome can be "Head" or "Tail".

```python
# WHat is the probability of getting 4 while rolling a dice.

# Total Number Of Outcome = 6       (1,2,3,4,5,6)
# Favorable Outcome = 1   (Getting a 4)
# Prob: 1/6
```

### **Marginal Probability**

It is the probability of single event happening without considering any other event.

```python
# You are having a bag full of balls: 5 red balls and 3 blue balls. What are the chances of picking
# red out of total balls.


# Favorable Outcome: 5 (Count of red balls)
# Total Number of Outcome: 8 (5 red balls + 3 blue balls)
# Prob: 5/8
```

```python
# There are total 100 students, out of them, 40 are studying mathematics, 30 are studying science and 30 are studying arts.
# A student has been selected randomly, Calculate the probability that the student does not belong to mathematics background.
```

```python
# Probability of not Mathematics
# Probability of picking a student from Mathematics background
# Fav Outome: 40 (Students from Mathematics background)
# Total Outcome: 100 (Total number of students)

# P(Mathematics) = 40/100  => 2/5

# P(Not Mathematics) = 1 - P(Mathematics)
# P(NT) = 1 - 2/5
# P => 3/5
```

### **Joint Probability**

It is the probability of two or more events happening at the same time. It's like what are the chances of this and that happening at the same time.

```python
# What will be the probability of getting a head on both the coins when two coins are tossed together.


# Favorable Outcome: 1(H,H)    (Getting Head on both the coins)
# Total Outcome: 4 (H,T) (H,H) (T,H) (T,T)

# Prob: 1/4


# To calculate joint probability
# Let's say there are two coins A and B
#  P(A) = 1/2   (Getting Head)
#  P(B) = 1/2   (Getting Head)
# P(A and B) = P(A) * P(B)
# P(A and B) = 1/2 * 1/2 => 1/4
```

```python
# If you roll 2 dice at the same time what will be the probability of both the dice showing 5.
# P(A) = 1/6 (Getting 5 on dice 1)
# P(B) = 1/6 (Getting 5 on dice 2)
# P(A and B) = 1/6*1/6 => 1/36

```

```python
# A coin is tossed and a dice is rolled.
# What is the probability of getting Heads and 4 on a dice.

# P(A) = 1/2  #(Getting head while tossing coin)
# P(B) = 1/6  #(Getting 4 while rolling a dice)

# P(A and B) = 1/2 * 1/6 => 1/12.
```

### **Conditional Probability**

It is the probability of an event happeing/occuring given that another event has already occured.

```python
# Suppose you know that a card drawn from a deck is a face card.
# What is the probability that the card is a King?

# Total Number of Outcome: 12 (Total number of face cards)
# Number of fav outcome: 4 (4 number of kings)
# Prob: 4/12 => 1/3.
```

```python
# A bag contains, 3 red, 4 green and 3 blue balls. Given that a ball drawn is not red.
# What is the probability that the ball is green.
# Favorable Outcome: 4 (4 green balls)
# Total Outcome: Not Red => It means green or blue balls => (4+3) = 7
# Prob: 4/7
```

```python
# A dice is rolled, what is the probability that the number is prime, given that
# it is greater than 2.

# Favorable Outcome = 2 (3,5)
# Total Outcome = 4 (3,4,5,6)

# P(Prime) = 2/4 => 1/2.
```
