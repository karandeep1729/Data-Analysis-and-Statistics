Create bar plots for the following, make sure all the plots appear in the same plane with x-labels and y-labels evenly spaced.

a. Create a bar plot that shows the visual representation of hp, mpg, weight, time-to-60 with respect to the number of cylinders for each of the brands.

```python
fig = plt.subplots(2,2,figsize=(20,10))
plt.subplot(2,2,1)
sns.barplot(x=data2['cylinders'],y=data2['hp'],data=data2,hue=data2['brand'])
plt.xlabel('Cylinders')
plt.ylabel('Horse Power')
plt.title("Cylinders v/s Horse Power")

plt.subplot(2,2,2)
sns.barplot(x=data2['cylinders'],y=data2['mpg'],data=data2,hue=data2['brand'])
plt.xlabel('Cylinders')
plt.ylabel('Miles per Gallon')
plt.title("Cylinders v/s Miles Per gallon")


plt.subplot(2,2,3)
sns.barplot(x=data2['cylinders'],y=data2['weightlbs'],data=data2,hue=data2['brand'])
plt.xlabel('Cylinders')
plt.ylabel('Weight LBS')
plt.title("Cylinders v/s Weight LBS")


plt.subplot(2,2,4)
sns.barplot(x=data2['cylinders'],y=data2['time-to-60'],data=data2,hue=data2['brand'])
plt.xlabel('Cylinders')
plt.ylabel('0-60')
plt.title("Cylinders v/s Time to 60")

plt.show()

#These dark lines black lines on top are knows as error bars
```
