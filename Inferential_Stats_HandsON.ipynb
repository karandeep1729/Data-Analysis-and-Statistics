# Let's suppose it was observed that the average purchase made by the Men of age group 18-25 was 10000. Is it still the same?

```python
data['Age'].unique()
```

```python
# Note down the null and alternate hypothesis
# Create a new or sub dataset based on the conditions.
# Now the new dataset will be my entire population.
# From this population we need to create sample.
# We perform hypothesis testing
# We conclude whether it is null or alternative.
```

```python
# Null Hypothesis -> Average purchase made by Men of the age group 18-25 is 10000.
# Alternate Hypothesis -> Average purchase made by Men of the age group 18-25 is not 10000.
```

```python
new_data = data.loc[(data['Age']==1) & (data['Gender']==1)]   #age==1 I am selecting all the rows which are of age grop 18-25
```

```python
new_data                            #Population
```

```python
#rule is to take 30 but practically when you will use t-test you will be taking 10 -20 % as sample and performing t-test
```

```python
# Whenever in problem statement there is mean or average mentioned we use t-test.
# Whenever there is percentage or preportion mentioned you have to use z-test.
```

```python
data.sample(5,random_state=12345)    #sample function is used to generate samples randomly
                                     #random_state is used to lock the pattern of your data.
                    #From 0 till infinite any +ve numeric value you can pass in random_state
```

```python
sample_size = 1000    # Approx 10% of my data I am selecting as sample out of population.
new_data_sample = new_data.sample(sample_size, random_state=0)
new_data_sample
```

```python
from scipy.stats import ttest_1samp     #I am importing one sample ttest from scipy. statistics library
```

```python
t_stat, p_value = ttest_1samp(new_data_sample['Purchase'],10000)
```

```python
p_value
```

```python
if p_value < 0.05:
  print("We have to reject null Hypothesis - Average purchase made by the Men is not 10000")
else:
  print("Failed to reject the null Hypothesis - Average purchase made by Men is 10000")


  # p_value > 0.05 Fail to reject null hypothesis.
```

```python
# Is the percentage of men who have spent more than 10000 the same for the ages 18-25 and 26-35?


# Null hypothesis -> Men who have spent more than 10000 are same for the ages 18-25 and 26-35.
# Alternate hypothesis ->Men who have spent more than 10000 are not same for the ages 18-25 and 26-35.
```

```python
df_new = data[data['Purchase']>10000]
```

```python
df_new
```

```python
data_men_1 = df_new.loc[(df_new['Gender']==1) & (df_new['Age']==2)]   # 18-25
data_men_2 = df_new.loc[(df_new['Gender']==1) & (df_new['Age']==3)]   # 26-35
```

```python
data_men_1
```

```python
data_men_2
```

```python
data_men_1.shape     # 1600
```

```python
data_men_2.shape     # 700
```

```python
sample_size = 1100
data_men_sample = data_men_1.sample(sample_size,random_state=500)   # From 0 to infinite any value.
data_men_2_sample = data_men_2.sample(sample_size,random_state=31)
```

```python
data_men_sample
```

```python
data_men_2_sample
```

```python
from scipy.stats import ttest_ind  #Two tailed independent sample t-test
```

```python
t_stat, p_value = ttest_ind(data_men_sample['Purchase'],data_men_2_sample['Purchase'],equal_var=True)
p_value
```

```python
if p_value < 0.05:
  print("We have to reject null Hypothesis - Men who have spent more than 10000 are not same for the ages 18-25 and 26-35")
else:
  print("Failed to reject the null Hypothesis - Men who have spent more than 10000 are same for the ages 18-25 and 26-35")
```

# Is the average purchase made by men and women of the age 18-25 is same

```python
# Null -> Average purchase made by men and women of the age 18-25 is same.
# Alternate -> Average purchase made my men and women of the age 18-25 is not same.
```

```python
# what we are comparing  -> Average
# If you see average -> t-test
# I am comparing here two groups
# 2 sample t-test
```

```python
data_men = data.loc[(data['Gender']==1) & (data['Age']==1)]
data_women = data.loc[(data['Gender']==0) & (data['Age']==1)]
```

```python
data_women
```

```python
data_men.shape   # 1000 approx 10%   Sample size must be same for both the datasets.
```

```python
data_women.shape   # 300 approx 10%
```

```python
sample_size = 500   #In between value we will select
data_men_sample = data_men.sample(sample_size,random_state=29)
data_women_sample = data_women.sample(sample_size,random_state=31)
```

```python
data_women_sample
```

```python
from scipy.stats import ttest_ind    #two tailed independent t-test

# Whenever we apply two sample t-test we need to consider variance of both the data is equal
```

```python
t_stat, p_value = ttest_ind(data_men_sample['Purchase'],data_women_sample['Purchase'],equal_var=True)
p_value
```

```python
if p_value < 0.05:
  print("We have to reject null Hypothesis - Average purchase made by the Men and Women is not equal")
else:
  print("Failed to reject the null Hypothesis - Average purchase made by Men and Women is equal")
```
